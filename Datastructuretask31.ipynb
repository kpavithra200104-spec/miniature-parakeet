```python
numbers=[13,24,67,38,89,46]
largest=second_largest=numbers[0]
for number in numbers:
    if number>largest:
        second_largest=largest
        largest=number
    elif number>second_largest and number!=largest:
        second_largest=number
print("Second largest number:",second_largest)
```

```python
numbers=[12,34,23,67,89,46,35]
reversed_list=[]
for i in range(len(numbers)-1,-1,-1):
    reversed_list.append(numbers[i])
print("Original list:",numbers)
print("Reversed list:",reversed_list)
```

```python
cart=[]
while True:
  print("1.Add item")
  print("2.View cart")
  print("3.Exit")

  choice=input("Enter your choice:")

if choice==1:
    item=input("Enter item name:")
    cart.append(item)
    print(item,"has been added to the cart.")

elif choice==2:
    product=input("Enter product to remove")
    if product in cart:
        cart.remove(product)
        print(product,"has been removed from the cart.")
 else:
        print(product,"is not in the cart.")

 elif choice=="3":
      print("Thank you for shopping with us!")
 else:
      print("Invalid choice.Please try again.")
```
