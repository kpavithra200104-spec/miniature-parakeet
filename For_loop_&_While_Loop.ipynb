# While Loop in python

### **While loop repeatedly executes a block of code as long as a given condition remains True**.
### **It is useful when the number of iterations is not known in advance and depends on dynamic conditions**.

### **Forloop used for iterating over a sequence (like a list, tuple, string, or range)**
### **It is well-suited for scenarios where the number of iterations is known or**
### **can be determined by the length of the sequence**

### Count Vowels 

```python
text = input("Enter a string: ")
vowels = "aeiouAEIOU"
count = 0
index = 0

while index < len(text):
    if text[index] in vowels:
        count += 1
    index += 1

print(f"Number of vowels in the string: {count}")
```

### Nested While 

```python
print(1,'* 10','=',10*1)
```

```python
table=1
while table<=10:
    multiplier=1
    while multiplier<=10:
        print(f"{table}*{multiplier}={table*multiplier}")
        multiplier=multiplier+1
    print('\n')
    table=table+1
```

```python
i=1
while i<=10:
    j=1
    while j<=10:
        print(i,'*',j,'=',i*j)
        j=j+1
    i=i+1
    print('\n')
```

### Find all even Numbers in the list

```python
numbers = [10, 15, 22, 33, 40, 55]

for number in numbers:
    if number % 2 == 0:
        print(number)
```

### Iterating over string 

```python
text = "Hello, World!"

for char in text:
    print(char)
```

```python
for i in range(1,11):
    for j in range(1,11):
        print(f"{i}*{j}={i*j}")
    print("\n")
```

### Flatten the list

```python
list_of_lists = [[1, 2, 3], [4, 5, 6], [7, 8, 9]] 


flattened_list = []
for i in list_of_lists:  #[1,2,3]
    for j in i:
        flattened_list.append(j)

print(f"Flattened list: {flattened_list}")
```

```python
def count(child_dict, i):
    if i not in child_dict.keys(): 
        return 1 
    ans = 1 
    for j in child_dict[i]: 
        ans += count(child_dict, j) 
        return ans 
child_dict = dict() 
    
    
child_dict[0] = [1,2] 
child_dict[1] = [3,4,5] 
child_dict[2] = [6,7,8] 

print(count(child_dict,0))
```
