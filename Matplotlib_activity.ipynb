```python
import matplotlib.pyplot as plt

# Data
x = [1, 2, 3, 4, 5]
y = [10, 20, 15, 25, 30]

# Create line plot
plt.plot(x, y, marker='o', color='blue', linestyle='-')

# Add labels and title
plt.xlabel('X-axis')
plt.ylabel('Y-axis')
plt.title('Simple Line Plot')

# Display grid and plot
plt.grid(True)
plt.show()
```

```python
import matplotlib.pyplot as plt

# Data
x = ['A', 'B', 'C', 'D', 'E']
y = [10, 25, 15, 30, 20]

# Create bar chart
plt.bar(x, y, color='skyblue')

# Add title and labels
plt.title('Simple Bar Chart')
plt.xlabel('Categories')
plt.ylabel('Values')

# Display grid and chart
plt.grid(axis='y', linestyle='--', alpha=0.5)
plt.show()
```

```python
import matplotlib.pyplot as plt

# Data
x = [1, 2, 3, 4, 5]
y = [10, 25, 15, 30, 20]

# Create scatter plot
plt.scatter(x, y, color='blue', marker='o')

# Add title and labels
plt.title('Simple Scatter Plot')
plt.xlabel('X-axis')
plt.ylabel('Y-axis')

# Display grid and plot
plt.grid(True)
plt.show()
```

```python
import matplotlib.pyplot as plt

# Data
data = [10, 20, 20, 30, 30, 30, 40, 40, 50, 50, 60, 70]

# Create histogram
plt.hist(data, bins=5, color='skyblue', edgecolor='black')

# Add title and labels
plt.title('Simple Histogram')
plt.xlabel('Values')
plt.ylabel('Frequency')

# Display grid and histogram
plt.grid(axis='y', linestyle='--', alpha=0.5)
plt.show()
```

```python
import matplotlib.pyplot as plt

# Data
labels = ['Python', 'Java', 'C++', 'JavaScript']
sizes = [40, 25, 20, 15]
colors = ['skyblue', 'lightgreen', 'orange', 'pink']

# Create pie chart
plt.pie(
    sizes,
    labels=labels,
    colors=colors,
    autopct='%1.1f%%',
    startangle=90
)

# Add title
plt.title('Programming Language Usage')

# Display pie chart
plt.show()
```

```python
import matplotlib.pyplot as plt

# Data
data = [10, 20, 25, 30, 35, 40, 45, 50, 60, 70, 100]

# Create box plot
plt.boxplot(data)

# Add title and labels
plt.title('Simple Box Plot')
plt.ylabel('Values')

# Display the plot
plt.grid(True)
plt.show()
```
