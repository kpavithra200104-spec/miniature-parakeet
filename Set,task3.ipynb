```python
my_list=[20,24,77,55,89]
my_set=set(my_list)
print("Original list:",my_list)
print("Set without duplicates:",my_set)
```

```python
set1={"Arun","Kavin","Ravi","Priya"}
set2={"Kavin","Priya","Suresh","Arun"}

common_students=set1.intersection(set2)
print("Common students:",common_students)
```

```python
class1={"Arun","Kavin","Ravi","Priya"}
class2={"Kavin","Priya","Suresh","Arun"}
unique_students=class1.union(class2)
print("All unique students:",unique_students)
```

```python
set1={1,2,3}
set2={1,2,3,4,5}
if set1.issubset(set2):
 print("Set1 is a subset of  Set2")
else:
 print("Set1 is not a subset of Set2")
```

```python
set1={1,2,3}
set2={4,5,6}
if set1.isdisjoint(set2):
  print("The sets are completely different")
else:
  print("The sets have common elements")
```

```python
numbers=set(range(1,21))
for num in numbers.copy():
  if num%2==0:
    numbers.remove(num)
print("Set after removing even numbers:",numbers)
```

```python
set1={1,2,3,4,5}
set2={4,5,6,7,8}
print("Union:",set1.union(set2))
print("Intersection:",set1.intersection(set2))
print("Difference(Set1-Set2):",set1.difference(set2))
print("Difference(Set2-Set1):",set2.difference(set1))
```
