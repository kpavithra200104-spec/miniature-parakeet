```python
library_books = ["The Hobbit", "1984", "To Kill a Mockingbird", "Pride and Prejudice"]

print("Current books in the library:")
print(library_books)
want_book = input("Do you want to take a book from the library? (yes/no): ").strip().lower()

if want_book == "yes":
    has_card = input("Do you have a library card? (yes/no): ").strip().lower()

    if has_card == "yes":
        chosen_book = input("Which book do you need to take?: ").strip()

        if chosen_book in library_books:
            library_books.remove(chosen_book)
            print(f"Success! You have checked out '{chosen_book}'.")
            print("Remaining books in the library:")
            print(library_books)
        else:
            print("Sorry, that book is not available in the library.")

    else:
        print("Sorry, you need a library card to check out a book.")

else:
    print("Okay, thank you for visiting the library!")
```
