```python
for n in range(2,101):
  prime=True

  for i in range(2,n):
    if n%i==0:
      prime=False
      break

  if prime:
    print(n)
```

```python
numbers=[10,25,7,45,18,32]

largest=numbers[0]

for i in numbers:
  if i>largest:
    largest=i

    print("Largest number=",largest)
```

```python
numbers=[10,15,20,25,30,35,40]

even=0
odd=0

for i in numbers:
  if i%2==0:
    even+=1
else:
    odd+=1

print("Even numbers=",even)
print("Odd numbers=",odd)
```

```python
numbers=[10,-5,0,20,-8,0,15]

positive=negative=zero=0

for num in numbers:
  if num>0:
    positive=positive+1
  if num<0:
    negative=negative+1
  if num==0:
    zero=zero+1

  print("Positve numbers=",positive)
  print("Negative numbers=",negative)
  print("Zeros=",zero)
```
