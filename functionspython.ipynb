```python
# The Following code checks if the user-provided number is prime or not

# Defining the function
# Logic : A number is not prime if :
# 		1. it is less than equal to 1 and
# 		2. it is divisible by any number except 1 and itself

def is_prime(number):
  if number > 1:
    for i in range(2,number):
      if (number % i) == 0:
        return False
    return True
  else:
    return False

# Taking user input
number = int(input("Enter a number to check: "))

# Condition Checking
if is_prime(number):
  print(f"{number} is a prime number.")
else:
  print(f"{number} is not a prime number.")
```

```python
def calculator(num1, num2, operator):
    """
    Performs basic arithmetic operations on two numbers.
    Supported operators: '+', '-', '*', '/'
    """
    if operator == '+':
        return num1 + num2
    elif operator == '-':
        return num1 - num2
    elif operator == '*':
        return num1 * num2
    elif operator == '/':
        if num2 == 0:
            return "Error: Cannot divide by zero!"
        return num1 / num2
    else:
        return "Error: Invalid operator!"

print(calculator(10, 5, '+'))   # Outputs: 15
print(calculator(10, 5, '-'))   # Outputs: 5
print(calculator(10, 5, '*'))   # Outputs: 50
print(calculator(10, 0, '/'))   # Outputs: Error: Cannot divide by zero!
print(calculator(10, 2, '%'))   # Outputs: Error: Invalid operator!
```
