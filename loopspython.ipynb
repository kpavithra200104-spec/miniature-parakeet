```python
n = 6
if n < 0:
    print("Factorial is not defined for negative numbers")
else:
    f = 1
    for i in range(1, n+1):
        f *= i
    print(f)
```

```python
n = 153
t = n
p = len(str(n))
s = 0

while t > 0:
    d = t % 10
    s += d ** p
    t //= 10

if s == n:
    print("Armstrong Number")
else:
    print("Not an Armstrong Number")
```

```python
rows = 5  # Number of rows

for i in range(1, rows + 1):  # Loop for rows
    for j in range(i):  # Loop for columns (stars)
        print("*", end=" ")  # Print star
    print()  # Move to the next line
```

```python
num = 12345

# One-liner solution
digit_sum = sum(map(int, str(num)))

print(f"The sum of digits is: {digit_sum}")  # Output: 15
```

```python
total_sum = 0

print("Please enter 10 numbers:")

for i in range(1, 11):
    num = float(input(f"Enter number {i}: "))
    total_sum += num  # Add the number directly to the total

average = total_sum / 10

print(f"\nThe average of the 10 numbers is: {average}")
```

```python
# Print a multiplication grid from 1 to 10
for i in range(1, 11):       # Outer loop: controls the rows
    for j in range(1, 11):   # Inner loop: controls the columns
        # Calculate product and format it to take up 4 spaces
        print(f"{i * j:4}", end="")
    print()                  # Moves to the next line after completing a row
```
