```python
import pandas as pd
import matplotlib.pyplot as plt
```

```python
from sklearn.datasets import load_iris

iris = load_iris()

df = pd.DataFrame(iris.data, columns=iris.feature_names)
df['species'] = iris.target

print(df.head())
```

```python
# Display first 5 rows
print(df.head())

# Display dataset information
print(df.info())

# Statistical summary
print(df.describe())

# Check missing values
print(df.isnull().sum())

# Check dataset shape
print(df.shape)

# Species count
print(df['species'].value_counts())
```

```python
df['species'].value_counts().plot(kind='bar')

plt.title('Iris Species Count')
plt.xlabel('Species')
plt.ylabel('Count')
plt.show()
```

```python
plt.scatter(df['sepal length (cm)'],
            df['petal length (cm)'])

plt.title('Sepal Length vs Petal Length')
plt.xlabel('Sepal Length')
plt.ylabel('Petal Length')
plt.show()
```

```python
df['petal length (cm)'].plot(kind='hist', bins=10)

plt.title('Petal Length Distribution')
plt.xlabel('Petal Length')
plt.show()
```

```python
import numpy as np
import matplotlib.pyplot as plt
from sklearn.datasets import load_iris
```

```python
iris = load_iris()

X = iris.data
y = iris.target

print("Iris Dataset")
print(X[:5])

print("Shape:", X.shape)
print("Feature Names:", iris.feature_names)
print("Target Names:", iris.target_names)
```

```python
print("Mean of Features:")
print(np.mean(X, axis=0))
```

```python
print("Median of Features:")
print(np.median(X, axis=0))
```

```python
print("Standard Deviation:")
print(np.std(X, axis=0))
```

```python
print("Minimum Values:")
print(np.min(X, axis=0))

print("Maximum Values:")
print(np.max(X, axis=0))
```

```python
print("Variance:")
print(np.var(X, axis=0))
```

```python
print("Correlation Matrix:")
print(np.corrcoef(X, rowvar=False))
```

```python
species_count = np.bincount(y)

print("Species Count:")
print(species_count)
```

```python
plt.bar(iris.target_names, species_count)

plt.title("Iris Species Count")
plt.xlabel("Species")
plt.ylabel("Count")
plt.show()
```

```python
plt.scatter(X[:, 0], X[:, 2], c=y)

plt.title("Sepal Length vs Petal Length")
plt.xlabel("Sepal Length (cm)")
plt.ylabel("Petal Length (cm)")
plt.show()
```

```python
plt.hist(X[:, 2], bins=10)

plt.title("Petal Length Distribution")
plt.xlabel("Petal Length (cm)")
plt.ylabel("Frequency")
plt.show()
```
