```python
n=int(input("Enter a number:"))

reverse=0

while n>0:
  digit=n%10
  reverse=reverse*10+digit
  n=n//10

  print("Reverse=",reverse)
```

```python
n=int(input("Enter a number"))

original=n
reverse=0

while n>0:
  digit=n%10
  reverse=reverse*10+digit
  n=n//10

  if original==reverse:
    print("Palindrome")
  else:
      print("Not a palindrome")
```
